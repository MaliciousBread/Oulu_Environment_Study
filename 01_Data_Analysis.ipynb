```
import pandas as pd
import numpy as np

print(f"Pandas : {pd.__version__} | Numpy : {np.__version__}")

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.2f}".format)

print("Chargement et fusion des datasets...")
df_sensor = pd.read_csv('Projet_Data/application.csv')
df_network = pd.read_csv('Projet_Data/lora.csv')

df_sensor['time'] = pd.to_datetime(df_sensor['time'], unit='ms')
df_network['time'] = pd.to_datetime(df_network['time'], unit='ms')

df = pd.merge(df_sensor, df_network, on=['time', 'deveui'], how='inner')
print(f"Dimensions du dataset fusionné : {df.shape}")

display(df.head())
```

```
df.info()

numeric_columns = df.select_dtypes(include=["float64", "int64", "int32"]).columns
print(f"\nNombre de colonnes numériques : {len(numeric_columns)}")
print("Liste complète des colonnes :", df.columns.tolist())
```

```
import missingno as msno;

print(f"Nombre de lignes dupliquées : {df.duplicated().sum()}\n")

missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

missing_data = pd.DataFrame({'Nombre': missing_count, 'Pourcentage (%)': missing_percent})
missing_data = missing_data[missing_data['Nombre'] > 0].sort_values(by='Pourcentage (%)', ascending=False)

print("Bilan des valeurs manquantes :")
display(missing_data)
msno.matrix(df)
```

```
display(df.describe().T)

zero_count = (df == 0).sum()
print("\nTop 15 des colonnes contenant le plus de zéros :")
display(zero_count.sort_values(ascending=False).head(15))
```

```
df = df.sort_values("time")

time_diff = df["time"].diff()

print("Type de la variable temporelle :", df["time"].dtype)
print("\nIntervalles de temps les plus fréquents (Fréquence d'envoi LoRaWAN) :")
print(time_diff.value_counts().head(10))
```
