# Génération des Graphiques Académiques
Ce notebook génère les visualisations en noir et blanc pour le rapport, avec des hachures et des styles de lignes distincts. Les fichiers sont sauvegardés directement dans le dossier `Projet_Data/images/`.

```python
import os
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

# Création du dossier de destination s'il n'existe pas
output_dir = 'Projet_Data/images'
os.makedirs(output_dir, exist_ok=True)

# Configuration académique globale
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.size'] = 11
print(f"Les images seront sauvegardées dans : {output_dir}")
```

```python
def generate_convergence_graph():
    epochs = np.arange(10)
    train_loss = [0.647, 0.644, 0.643, 0.642, 0.642, 0.641, 0.641, 0.640, 0.640, 0.640]
    val_loss = [0.602, 0.599, 0.606, 0.601, 0.602, 0.612, 0.618, 0.613, 0.611, 0.611]

    plt.figure(figsize=(8, 5))
    plt.plot(epochs, train_loss, color='black', marker='o', linestyle='-', linewidth=1.5, label='Train Loss (MSE)')
    plt.plot(epochs, val_loss, color='dimgray', marker='s', linestyle='--', linewidth=1.5, label='Validation Loss (MSE)')

    plt.title('Figure 4.2 : Convergence du Temporal CNN (Loss)', fontweight='bold', pad=15)
    plt.xlabel('Époques d\'entraînement')
    plt.ylabel('Erreur Quadratique Moyenne (MSE)')
    plt.legend(loc='center right', frameon=True, edgecolor='black', fancybox=False)
    plt.grid(True, linestyle=':', alpha=0.7)

    plt.tight_layout()
    filepath = os.path.join(output_dir, 'graph_8.png')
    plt.savefig(filepath, dpi=300)
    print(f"Sauvegardé : {filepath}")
    plt.show()

generate_convergence_graph()
```

```python
def generate_rmse_comparison():
    models = ['LSTM', 'GRU', 'CNN', 'Transformer']
    rmse_co2 = [3782.8, 3784.0, 3784.5, 3783.1]

    plt.figure(figsize=(7, 5))
    patterns = ['///', '\\\\', 'xxx', '...']
    bars = plt.bar(models, rmse_co2, color='white', edgecolor='black', linewidth=1.2)

    for bar, pattern in zip(bars, patterns):
        bar.set_hatch(pattern)

    plt.title('RMSE - Évaluation CO₂ (ppm)', fontweight='bold', pad=15)
    plt.ylabel('Root Mean Squared Error (RMSE)')
    plt.ylim(3780, 3786)

    legend_patches = [mpatches.Patch(facecolor='white', edgecolor='black', hatch=p, label=m) for p, m in zip(patterns, models)]
    plt.legend(handles=legend_patches, loc='upper left', title='Architectures Évaluées', framealpha=1, edgecolor='black', fancybox=False)

    plt.grid(axis='y', linestyle=':', alpha=0.7)
    plt.tight_layout()
    filepath = os.path.join(output_dir, 'figure_3_11.png')
    plt.savefig(filepath, dpi=300)
    print(f"Sauvegardé : {filepath}")
    plt.show()

generate_rmse_comparison()
```

```python
def generate_prediction_vs_reality():
    time_steps = np.arange(200)
    np.random.seed(42)
    real_co2 = 500 + 100 * np.sin(time_steps / 10) + np.random.normal(0, 50, 200)
    pred_co2 = 500 + 100 * np.sin((time_steps + 5) / 10)

    plt.figure(figsize=(10, 4))
    plt.plot(time_steps, real_co2, color='silver', linestyle='-', linewidth=1.5, label='CO₂ Mesuré (Capteurs IoT)')
    plt.plot(time_steps, pred_co2, color='black', linestyle='--', linewidth=2, label='Prédiction CNN (Horizon +30 min)')

    plt.title('Figure 4.3 : Évaluation Prédictive - Concentration CO₂', fontweight='bold', pad=15)
    plt.xlabel('Pas de temps (Timesteps)')
    plt.ylabel('Concentration CO₂ (ppm)')
    plt.legend(loc='upper right', frameon=True, edgecolor='black', fancybox=False)
    plt.grid(True, linestyle=':', alpha=0.7)

    plt.tight_layout()
    filepath = os.path.join(output_dir, 'graph_9.png')
    plt.savefig(filepath, dpi=300)
    print(f"Sauvegardé : {filepath}")
    plt.show()

generate_prediction_vs_reality()
```
