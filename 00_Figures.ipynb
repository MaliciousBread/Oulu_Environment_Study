```
import os
import matplotlib.pyplot as plt

os.makedirs("Projet_Data/images", exist_ok=True)

fig, ax = plt.subplots(figsize=(14, 6), dpi=300)

zones = [
    (400, 600, "#2ecc71", "Air extérieur / Excellent", "(400 - 600 ppm)", -1),
    (600, 1000, "#f1c40f", "Acceptable / Conforme", "(600 - 1000 ppm)", 1),
    (1000, 1500, "#e67e22", "Seuil de somnolence", "(1000 - 1500 ppm)", -1),
    (1500, 2500, "#e74c3c", "Air vicié / Céphalées", "(1500 - 2500 ppm)", 1),
    (2500, 5000, "#8e44ad", "Critique / Léthargie", "(> 2500 ppm)", -1),
]

# Draw the base horizontal bar
for xmin, xmax, color, _, _, _ in zones:
    ax.barh(0, xmax - xmin, left=xmin, height=0.28, color=color, edgecolor="black", linewidth=1.2)

# Place labels alternating top and bottom with callout lines
for xmin, xmax, _, title, sub, direction in zones:
    x_mid = (xmin + xmax) / 2
    y_bar_edge = 0.14 if direction == 1 else -0.14
    y_text = 0.42 if direction == 1 else -0.42

    ax.plot([x_mid, x_mid], [y_bar_edge, y_text * 0.75], color="#2c3e50", lw=1.2, linestyle="--")
    
    label_str = f"{title}\n{sub}"
    ax.text(
        x_mid,
        y_text,
        label_str,
        ha="center",
        va="center",
        fontsize=9,
        fontweight="bold",
        color="#1a252f",
        bbox=dict(boxstyle="round,pad=0.35", facecolor="#f8f9fa", edgecolor="#bdc3c7", lw=1)
    )

ax.set_xlim(350, 5050)
ax.set_ylim(-0.75, 0.75)
ax.set_yticks([])
ax.set_xlabel("Concentration en CO₂ (parties par million - ppm)", fontsize=11, fontweight="bold", labelpad=10)
ax.set_title("Figure 1.3 : Seuils de concentration en CO₂ en milieu clos et impacts cognitifs", fontsize=12, fontweight="bold", pad=20)

plt.tight_layout()
plt.savefig("Projet_Data/images/figure_1_3.png", bbox_inches="tight")
plt.show()
print("Saved clean figure to Projet_Data/images/figure_1_3.png")
```
